```python
DAGSHUB_USERNAME = "danghoanghai042"
DAGSHUB_REPO     = "ppe_v2"
DAGSHUB_TOKEN    = userdata.get('ppe_v1')

os.environ["MLFLOW_TRACKING_USERNAME"] = DAGSHUB_USERNAME
os.environ["MLFLOW_TRACKING_PASSWORD"] = DAGSHUB_TOKEN
os.environ["DAGSHUB_USER_TOKEN"]       = DAGSHUB_TOKEN

TRACKING_URI = f"https://dagshub.com/{DAGSHUB_USERNAME}/{DAGSHUB_REPO}.mlflow"
os.environ["MLFLOW_TRACKING_URI"] = TRACKING_URI
mlflow.set_tracking_uri(TRACKING_URI)
```

```python
if mlflow.active_run():
    mlflow.end_run()

dagshub.init(
    repo_owner=DAGSHUB_USERNAME,
    repo_name=DAGSHUB_REPO,
    mlflow=True
)
```

```python
import pandas as pd
from pathlib import Path
import yaml

results_csv = Path("/content/yolov5/runs/train/yolov5l_ppe/results.csv")
hyp_yaml = Path("/content/yolov5/runs/train/yolov5l_ppe/hyp.yaml")
opt_yaml = Path("/content/yolov5/runs/train/yolov5l_ppe/opt.yaml")

df = pd.read_csv(results_csv)
df.columns = df.columns.str.strip()

metric_map = {
    "train/box_loss": "train/box_loss",
    "train/obj_loss": "train/obj_loss",
    "train/cls_loss": "train/cls_loss",

    "metrics/precision": "metrics/precisionB",
    "metrics/recall": "metrics/recallB",
    "metrics/mAP_0.5": "metrics/mAP50B",
    "metrics/mAP_0.5:0.95": "metrics/mAP50-95B",

    "val/box_loss": "val/box_loss",
    "val/obj_loss": "val/obj_loss",
    "val/cls_loss": "val/cls_loss",

    "x/lr0": "lr/pg0",
    "x/lr1": "lr/pg1",
    "x/lr2": "lr/pg2",
}

with open(hyp_yaml, 'r') as f:
  hyp = yaml.safe_load(f)

with open(opt_yaml, 'r') as f:
  opt = yaml.safe_load(f)

mlflow.set_experiment("PPE_Detection")

with mlflow.start_run(run_name="yolov5l_repo_v2"):
    for _, row in df.iterrows():
        step = int(row["epoch"])

        for src_name, dst_name in metric_map.items():
            if src_name in df.columns and pd.notna(row[src_name]):
                mlflow.log_metric(dst_name, float(row[src_name]), step=step)
    mlflow.log_params(hyp)
    mlflow.log_params(opt)
    mlflow.log_artifact(str(results_csv))
    mlflow.log_artifact('/content/yolov5/runs/train/yolov5l_ppe/weights')
    mlflow.log_artifact('/content/yolov5/runs/train/yolov5l_ppe/hyp.yaml')
    mlflow.log_artifact('/content/yolov5/runs/train/yolov5l_ppe/opt.yaml')
```
